```python
%sql
SELECT trade_date, open, close, ldcp, volume
FROM workspace.silver.daily_prices
WHERE symbol = 'LUCK' AND trade_date BETWEEN '2025-04-22' AND '2025-05-06'
ORDER BY trade_date
```

```python
%sql
SELECT sector_code, COUNT(DISTINCT symbol) AS stocks,
       MIN(trade_date) AS first_seen, MAX(trade_date) AS last_seen
FROM workspace.silver.daily_prices
GROUP BY sector_code ORDER BY sector_code
```

```python
%sql
SELECT sector_code, symbol, MAX(company_name) AS company_name, COUNT(*) AS days
FROM workspace.silver.daily_prices
WHERE sector_code IN ('36', '3610', '0839') OR sector_code IS NULL
GROUP BY sector_code, symbol
ORDER BY sector_code, days DESC
LIMIT 60
```
