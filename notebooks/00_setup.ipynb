```python
%sql
CREATE TABLE IF NOT EXISTS workspace.bronze.mkt_summary (
  trade_date       STRING,
  symbol           STRING,
  sector_code      STRING,
  company_name     STRING,
  open_raw         STRING,
  high_raw         STRING,
  low_raw          STRING,
  close_raw        STRING,
  volume_raw       STRING,
  ldcp_raw         STRING,
  extra_1          STRING,
  extra_2          STRING,
  extra_3          STRING,
  file_trade_date  DATE,
  source_file      STRING,
  batch_id         STRING,
  load_timestamp   TIMESTAMP
)
USING DELTA
PARTITIONED BY (file_trade_date)
```

```python
%sql

CREATE TABLE IF NOT EXISTS workspace.ops.pipeline_execution_logs (
    batch_id          STRING,
    layer             STRING,
    load_type         STRING,
    parameter         STRING,
    start_time        TIMESTAMP,
    end_time          TIMESTAMP,
    duration_seconds  DOUBLE,
    status            STRING,
    rows_inserted     BIGINT,
    rows_updated      BIGINT,
    rows_quarantined  BIGINT,
    error_message     STRING,
    load_timestamp    TIMESTAMP
);
```

```python
%sql
CREATE TABLE IF NOT EXISTS workspace.ops.quarantine_mkt_summary(
    raw_record STRING,
    reason STRING,
    layer STRING,
    source_file STRING,
    file_trade_date DATE,
    batch_id STRING,
    load_timestamp TIMESTAMP
)
```

```python
%sql
CREATE TABLE IF NOT EXISTS workspace.silver.daily_prices (
  symbol            STRING NOT NULL,
  trade_date        DATE   NOT NULL,
  sector_code       STRING,
  company_name      STRING,
  open              DOUBLE,
  high              DOUBLE,
  low               DOUBLE,
  close             DOUBLE,
  volume            BIGINT,
  ldcp              DOUBLE,
  is_traded         BOOLEAN,
  is_ohlc_valid     BOOLEAN,
  row_hash          STRING,
  source_file       STRING,
  bronze_batch_id   STRING,
  batch_id          STRING,
  load_timestamp    TIMESTAMP,
  updated_timestamp TIMESTAMP
) USING DELTA
```
